```python
# Assignment 4
# Z hypothesis testing
def z_test(sample_mean, population_mean, sigma, alpha=0.05):
    z_value = (sample_mean - population_mean) / sigma
    z_critical = 1.96  # for 95% confidence level

    print("Z-score:", z_value)

    if abs(z_value) > z_critical:
        print("Reject the null hypothesis")
        print("The claim is accepted")
    else:
        print("Accept the null hypothesis")
        print("The claim is rejected")

# Given values
sample_mean = 88
population_mean = 85
sigma = 1

z_test(sample_mean, population_mean, sigma)
```
