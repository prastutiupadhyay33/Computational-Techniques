```python
# Assignment 3
# Given confusion matrix values
tp = 199
fn = 1
fp = 3
tn = 247

def accuracy(tp,tn,fp,fn):
    return (tp + tn) / (tp + tn + fp + fn)

def precision(tp,fp):
    return tp / (tp + fp)

def recall(tp,fn):
    return tp / (tp + fn)

def f1_score(p,r):
    return 2 * p * r / (p + r)

# Calculate metrics
acc = accuracy(tp, tn, fp, fn)
prec = precision(tp, fp)
rec = recall(tp, fn)
f1 = f1_score(prec, rec)

# Print results
print("Accuracy :", round(acc))
print("Precision:", round(prec))
print("Recall   :", round(rec))
print("F1-Score :", round(f1))
```
